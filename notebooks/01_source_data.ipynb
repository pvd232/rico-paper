Our first step is to download the published K562 count matrix and verify its bytes against `references/sources.toml`. We will inspect the original download code together before writing this step.

Reusable download and validation code will live in `src/rico_paper`. This notebook will call that code and show what each operation produces.

```python
import shutil
import sys
from importlib.metadata import version
from pathlib import Path

import jax

PROJECT_ROOT = Path.home() / "rico-paper"

assert Path(sys.prefix).resolve() == (PROJECT_ROOT / ".venv").resolve(), (
    "Select the RICO paper (.venv) kernel."
)
assert sys.version_info[:2] == (3, 12)

ARIA2 = shutil.which("aria2c")
assert ARIA2 is not None, "The download requires aria2c."

print(f"Python: {sys.executable}")
for package in ("jax", "pydantic", "viper-provenance"):
    print(f"{package}: {version(package)}")
print(f"GPU: {[device.device_kind for device in jax.devices('gpu')]}")
print(f"Downloader: {ARIA2}")
```
